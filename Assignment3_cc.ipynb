```python
''' Q1. You are a data analyst at an e-commerce company. Your task is to analyze sales data,
clean and process it, perform statistical analysis, and generate visual insights to help the
business make informed decisions.
1. Use your roll number as the random seed to generate unique data for each student.
2. Follow the questions step by step, as each part builds upon the previous one.
3. Submit the assignment as a Jupyter Notebook (.ipynb) with:
• Well-documented code and clear comments.
• Markdown explanations for each step.
• Screenshots of the plots and outputs'''

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
ROLL_NO = 23
np.random.seed(ROLL_NO)
sales_data = np.random.randint(1000, 5000, size=(12,4))
sales_data
months = ['Jan','Feb','Mar','Apr','May','Jun',
          'Jul','Aug','Sep','Oct','Nov','Dec']

columns = ['Electronics','Clothing','Home & Kitchen','Sports']

df = pd.DataFrame(sales_data, columns=columns, index=months)
df
df.head()
df.describe()
total_category_sales = df.sum(axis=0)
total_month_sales = df.sum(axis=1)

print("Total Sales per Category:\n", total_category_sales)
print("\nTotal Sales per Month:\n", total_month_sales)
growth = df.pct_change()
avg_growth = growth.mean()

avg_growth
df['Total Sales'] = df.sum(axis=1)
df['Growth Rate'] = df['Total Sales'].pct_change()*100
df
if ROLL_NO % 2 == 0:
    df['Electronics'] *= 0.90   # 10% discount
else:
    df['Clothing'] *= 0.85      # 15% discount

df
```

```python
df[['Electronics','Clothing','Home & Kitchen','Sports']].plot(marker='o')
plt.title("Monthly Sales Trends")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['Electronics','Clothing','Home & Kitchen','Sports']])
plt.title("Sales Distribution by Category")
plt.show()
```

```python
'''Q.2 For the array: array = np.array([[1, -2, 3],[-4, 5, -6]])
i. Find element-wise absolute value
ii. Find the 25th, 50th, and 75th percentile of flattened array, for each column, for each
row.
iii. Mean, Median and Standard Deviation of flattened array, of each column, and
each row'''
array = np.array([[1,-2,3],
                  [-4,5,-6]])
```

```python
#i. Element-wise Absolute Value
np.abs(array)
```

```python
#ii. Percentiles
np.percentile(array.flatten(), [25,50,75])
```

```python
#Column-wise
np.percentile(array, [25,50,75], axis=0)
```

```python
#Row-wise
np.percentile(array, [25,50,75], axis=1)
```

```python
#Flattened
# iii. Mean, Median, Standard Deviation
np.mean(array)
np.median(array)
np.std(array)
```

```python
#Column-wise
np.mean(array, axis=0)
np.median(array, axis=0)
np.std(array, axis=0)
```

```python
#Row-wise
np.mean(array, axis=1)
np.median(array, axis=1)
np.std(array, axis=1)
```

```python
'''Q.4 WAP to swap two elements in a list in Python Using a Temporary Variable.'''
lst = [10,20,30,40]

i, j = 1, 3   # positions to swap

temp = lst[i]
lst[i] = lst[j]
lst[j] = temp

print(lst)
```

```python
#Q.5 Write a Python program to swap two elements in a set by converting it to a list first.
s = {10,20,30,40}

# convert set → list
lst = list(s)

# swap elements
lst[0], lst[1] = lst[1], lst[0]

# convert back to set
s = set(lst)

print(s)
```
